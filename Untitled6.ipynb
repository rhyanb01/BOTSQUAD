```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
np.random.seed(SEED)

def make_grid(rows=15, cols=15, obstacle_prob=0.2, seed=42):
    rng = np.random.default_rng(seed)
    grid = (rng.random((rows, cols)) < obstacle_prob).astype(int)
    grid[0, 0] = 0
    grid[rows - 1, cols - 1] = 0
    return grid

grid = make_grid()
print(grid)


def manhattan(p, q):
    return abs(p[0] - q[0]) + abs(p[1] - q[1])



def neighbors(grid, cell):
    r, c = cell
    rows, cols = grid.shape
    for dr, dc in [(-1,0), (1,0), (0,-1), (0,1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols and grid[nr, nc] == 0:
            yield (nr, nc)

def heuristic_hill_climb(grid, start, goal):
    current = start
    path = [current]
    visited = {current}
    expansions = 0

    while current != goal:
        expansions += 1
        h_cur = manhattan(current, goal)
        best_next = None
        best_h = h_cur

        for nb in neighbors(grid, current):
            if nb in visited:
                continue
            h_nb = manhattan(nb, goal)
            if h_nb < best_h:
                best_h = h_nb
                best_next = nb

        if best_next is None:  # stuck
            return {
                "found": False,
                "path": path,
                "cost": len(path) - 1,
                "expansions": expansions,
            }

        visited.add(best_next)
        path.append(best_next)
        current = best_next

    return {
        "found": True,
        "path": path,
        "cost": len(path) - 1,
        "expansions": expansions,
    }

def plot_grid_path(grid, result, start, goal, title):
    fig, ax = plt.subplots(figsize=(4, 4))
    ax.imshow(grid, cmap="gray_r")
    ax.scatter(start[1], start[0], c="green", s=100, label="start")
    ax.scatter(goal[1], goal[0], c="red", s=100, marker="*", label="goal")
    if result["found"]:
        xs = [p[1] for p in result["path"]]
        ys = [p[0] for p in result["path"]]
        ax.plot(xs, ys, c="blue", linewidth=3)
    ax.set_title(f"{title}\nfound={result['found']}, cost={result['cost']}, exp={result['expansions']}")
    ax.set_xticks([])
    ax.set_yticks([])
    ax.legend()
    plt.tight_layout()
    plt.show()

grid = make_grid(seed=SEED)
start = (0, 0)
goal = (grid.shape[0]-1, grid.shape[1]-1)

res_hc = heuristic_hill_climb(grid, start, goal)
print(res_hc)

plot_grid_path(grid, res_hc, start, goal, title="Example 1: Heuristic (Hill-Climbing)")
```

```python
import heapq

# best-first search using the same helper functions
def best_first_search(grid, start, goal):
    frontier = []
    heapq.heappush(frontier, (manhattan(start, goal), start))
    came_from = {start: None}
    expansions = 0

    while frontier:
        _, current = heapq.heappop(frontier)
        expansions += 1

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = came_from[current]
            path.reverse()
            return {
                "found": True,
                "path": path,
                "cost": len(path) - 1,
                "expansions": expansions,
            }

        for nb in neighbors(grid, current):
            if nb not in came_from:
                came_from[nb] = current
                heapq.heappush(frontier, (manhattan(nb, goal), nb))

    return {
        "found": False,
        "path": [],
        "cost": float("inf"),
        "expansions": expansions,
    }

res_best = best_first_search(grid, start, goal)
print(res_best)

plot_grid_path(grid, res_best, start, goal, title="Example 2: Greedy Best-First")

print("== Direct comparison (same grid) ==")
print("Heuristic HC:", res_hc)
print("Best-first :", res_best)
```

```python
import networkx as nx
import heapq

SEED = 42

def make_random_graph(n_nodes=15, edge_prob=0.28, w_low=1, w_high=10, seed=42):
    rng = np.random.default_rng(seed)
    G = nx.Graph()
    for i in range(n_nodes):
        G.add_node(i)
    for i in range(n_nodes):
        for j in range(i + 1, n_nodes):
            if rng.random() < edge_prob:
                w = float(rng.integers(w_low, w_high + 1))
                G.add_edge(i, j, weight=w)
    return G

def dijkstra_shortest_path(G, start, goal):
    frontier = [(0.0, start)]
    dist = {start: 0.0}
    prev = {start: None}
    expansions = 0

    while frontier:
        d, u = heapq.heappop(frontier)
        expansions += 1
        if u == goal:
            break
        if d > dist[u]:
            continue
        for v in G.neighbors(u):
            w = G[u][v]["weight"]
            nd = d + w
            if v not in dist or nd < dist[v]:
                dist[v] = nd
                prev[v] = u
                heapq.heappush(frontier, (nd, v))

    if goal not in dist:
        return {"found": False, "path": [], "cost": float("inf"), "expansions": expansions}

    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = prev[cur]
    path.reverse()

    return {"found": True, "path": path, "cost": dist[goal], "expansions": expansions}

def plot_graph_path(G, pos, result, start, goal, title):
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(5, 5))
    nx.draw_networkx_nodes(G, pos, node_color="lightgray", ax=ax)
    nx.draw_networkx_labels(G, pos, ax=ax)
    nx.draw_networkx_edges(G, pos, edge_color="gray", ax=ax)
    edge_labels = nx.get_edge_attributes(G, "weight")
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=7, ax=ax)

    ax.scatter(pos[start][0], pos[start][1], c="green", s=200, label="start")
    ax.scatter(pos[goal][0], pos[goal][1], c="red", s=200, marker="*", label="goal")

    if result["found"]:
        path_edges = list(zip(result["path"], result["path"][1:]))
        nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color="blue", width=3, ax=ax)

    ax.set_title(f"{title}\nfound={result['found']}, cost={result['cost']:.1f}, exp={result['expansions']}")
    ax.axis("off")
    ax.legend()
    plt.tight_layout()
    plt.show()

G = make_random_graph(seed=SEED)
pos = nx.spring_layout(G, seed=SEED)
start_g = 0
goal_g = 14

res_dij = dijkstra_shortest_path(G, start_g, goal_g)
print(res_dij)

plot_graph_path(G, pos, res_dij, start_g, goal_g, title="Example 3: Dijkstra on Weighted Graph")
```

```python
import math
import matplotlib.pyplot as plt
import heapq

def a_star_search(G, start, goal, pos):
    def h(u):
        (x1, y1) = pos[u]
        (x2, y2) = pos[goal]
        return math.hypot(x1 - x2, y1 - y2)

    frontier = [(h(start), 0.0, start)]
    g_cost = {start: 0.0}
    prev = {start: None}
    expansions = 0

    while frontier:
        f, g, u = heapq.heappop(frontier)
        expansions += 1
        if u == goal:
            break
        if g > g_cost[u]:
            continue
        for v in G.neighbors(u):
            w = G[u][v]["weight"]
            new_g = g + w
            if v not in g_cost or new_g < g_cost[v]:
                g_cost[v] = new_g
                prev[v] = u
                f_new = new_g + h(v)
                heapq.heappush(frontier, (f_new, new_g, v))

    if goal not in g_cost:
        return {"found": False, "path": [], "cost": float("inf"), "expansions": expansions}

    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = prev[cur]
    path.reverse()

    return {"found": True, "path": path, "cost": g_cost[goal], "expansions": expansions}

res_astar = a_star_search(G, start_g, goal_g, pos)
print(res_astar)

plot_graph_path(G, pos, res_astar, start_g, goal_g, title="Example 4: A* on Same Graph")

print("==== Comparison on the same graph ====")
print("Dijkstra:", res_dij)
print("A*:", res_astar)
```
